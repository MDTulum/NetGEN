Question 3

part a

```r
hw1_data = read.csv("https://raw.githubusercontent.com/hgweon2/ss3859/master/hw1_data1.csv")
idx = which(hw1_data$x1 < 4)
a = length(idx)

cat("# of obs where x1 lower than 4: ", a)
```

part b

```r
idx = which((hw1_data$x1 < 4) & (hw1_data$x2 == "L"))
b = length(idx)
cat("# of obs where x1 lower than 4 & x2 is L: ", b)
```

part c

```r
idx = which(hw1_data$x2 == "L")
A = hw1_data[idx,]
cat("Mean:", mean(A$x1), "\n")
cat("median:", median(A$x1), "\n")
cat("sd:", sd(A$x1), "\n")
```

part d

H0 = 4 and H1 not equal to 4.

```r
x = hw1_data$x1
sample_mean = mean(x)
cat("Mean of X1 (x_bar): ", sample_mean, "\n")
sample_sd = sd(x)
n = length(x)
t_stat = (sample_mean - 4)/(sample_sd/sqrt(n))
p_val = 2*(1 - pt(abs(t_stat),df=n-1))
cat("p_val: ", p_val,"\n")
# we can do the same test using function t.test
t.test(x = x, mu = 4, alternative = c("two.sided"), conf.level = 0.95)
```

Since the p-value is 8.87%, which is more than significance level of 5%, then there is not enough evidence that the true mean of x1 differs from 4, so we fail to reject H0 of mean being 4.

part e

the same test as d but we use the subset A (from part c) of only those with x2=L and test against 2 instead of 4. the other change is that the hypothesis is about being greater than something so no two sided distribution any more and only one side (right side for greater) is considered

```r
idx = which(hw1_data$x2 == "L")
A = hw1_data[idx,]
x = A$x1
sample_mean = mean(x)
cat("Mean of X1 (x_bar): ", sample_mean, "\n")
sample_sd = sd(x)
n = length(x)
t_stat = (sample_mean - 2)/(sample_sd/sqrt(n))
p_val = 1 - pt(t_stat,df=n-1)
cat("p_val: ", p_val,"\n")
# we can do the same test using function t.test
t.test(x = x, mu = 2, alternative = c("greater"), conf.level = 0.95)
```

Since the p-value is 4.2%, which is less than the cut off of 5%, the statement that the true mean of x1 where x2 = L is greater than 2 is convincing.
